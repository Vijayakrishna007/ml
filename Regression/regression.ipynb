Life expectancy is the key metric for assessing population health. It tells us the average age of death in a population4. By definition, life expectancy is based on an estimate of the average age members of a particular population group will be when they die. Life expectancy depends on several factors, the most important being gender and birth year. Generally, females have a slightly higher life expectancy than males due to biological differences.

Filling null value with mean

```python
df['Life expectancy '].fillna(df['Life expectancy '].mean(), inplace=True)
df['Adult Mortality'].fillna(df['Adult Mortality'].mean(), inplace=True)
df['Alcohol'].fillna(df['Alcohol'].mean(), inplace=True)
df['Hepatitis B'].fillna(df['Hepatitis B'].mean(), inplace=True)
df[' BMI '].fillna(df[' BMI '].mean(), inplace=True)
df['Polio'].fillna(df['Polio'].mean(), inplace=True)
df['Total expenditure'].fillna(df['Total expenditure'].mean(), inplace=True)
df['Diphtheria '].fillna(df['Diphtheria '].mean(), inplace=True)
df['GDP'].fillna(df['GDP'].mean(), inplace=True)
df['Population'].fillna(df['Population'].mean(), inplace=True)
df[' thinness  1-19 years'].fillna(df[' thinness  1-19 years'].mean(), inplace=True)
df[' thinness 5-9 years'].fillna(df[' thinness 5-9 years'].mean(), inplace=True)
df['Population'].fillna(df['Population'].mean(), inplace=True)
df['Income composition of resources'].fillna(df['Income composition of resources'].mean(), inplace=True)
df['Schooling'].fillna(df['Schooling'].mean(), inplace=True)
```

 to fillsame type(int,str) all at once df = df.fillna(df.mode().iloc[0])

Checking for outliers

```python
# Using boxplots of all features
df.boxplot(figsize=(12,10))
plt.tick_params(axis='x', labelrotation = 90) #plt.xticks(rotation = 90)
plt.show()
```

there are some outliers so we should remove it

Deteccting the outliers

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
print(((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))))
print(((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).sum())
```

Removing the outliers

```python
df1 = df[~((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).any(axis=1)]
print(((df1 < (Q1 - 1.5 * IQR)) | (df1 > (Q3 + 1.5 * IQR))).sum())
print(df1)
```

cross checking by plottong box plot

```python
# Checking for outliers
# Using boxplots of all features
print(((df1 < (Q1 - 1.5 * IQR)) | (df1 > (Q3 + 1.5 * IQR))).sum())
df1.boxplot(figsize=(12,10))
plt.tick_params(axis='x', labelrotation = 90)
plt.show()
```

                                                            Simple Linear Regression

```python
data1=X_scaled.loc[:,['infant deaths','under-five deaths ']]
data1.head()
```

Here to perform slr i selected  under-five deaths as dependent variable and infant deaths as independent variable

plotting scatter plot

```python
data1.plot(x='infant deaths',y='under-five deaths ',style='o')
plt.xlabel('infant deaths')
plt.ylabel('under-five deaths')
plt.show()
```

making dataframes

```python
x=pd.DataFrame(data1['infant deaths'])
y=pd.DataFrame(data1['under-five deaths '])
```

Dividing training and testing data

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split (x, y, test_size=0.2, random_state=1)
```

printing shape of draing and test data

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

Fitting Training and test data

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(x_train,y_train)
```

Printing Regression Intercept

```python
print(reg.intercept_)
```

Printing Regression Coefficent

```python
print(reg.coef_)
```

predicting Y

```python
y_pred = reg.predict(x_test)
y_pred
```

Printing actual values

```python
y_test
```

plotting the Relationship between infant deaths and under-five deaths with best fit line

```python
data1.plot(x='infant deaths',y='under-five deaths ',style='o')
a, b = np.polyfit(x_test['infant deaths'], y_pred, 1)
plt.plot(x_test, a*x_test+b, color="red", linewidth=2)

plt.xlabel(" (infant deaths)")
plt.ylabel(" (under-five deaths )")
plt.title("Relationship between infant deaths andunder-five deaths " )
plt.show()
```

Model validation 

```python
from sklearn import metrics
from sklearn.metrics import r2_score
print("test")
print('Mean absolute error :',metrics.mean_absolute_error(y_test,y_pred))
print('mean sq error:',metrics.mean_squared_error(y_test,y_pred))
print("root mean square error:",np.sqrt(metrics.mean_squared_error(y_test,y_pred)))
print("R-squared score:",r2_score(y_test, y_pred))
```

```python
y_train_pred = reg.predict(x_train)
mse_train = metrics.mean_squared_error(y_train, y_train_pred)
print("Training MSE:",mse_train)
```

```python
sns.regplot(x=y_test,y=y_pred,ci=None,color ='red')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve
# Calculate the learning curve
train_sizes, train_scores, test_scores = learning_curve(reg, x, y)
# Calculate the mean and standard deviation of the training and test scores
train_scores_mean = np.mean(train_scores, axis=1)
train_scores_std = np.std(train_scores, axis=1)
test_scores_mean = np.mean(test_scores, axis=1)
test_scores_std = np.std(test_scores, axis=1)
# Plot the learning curve
plt.fill_between(train_sizes, train_scores_mean - train_scores_std,
                 train_scores_mean + train_scores_std, alpha=0.1,
                 color="r")
plt.fill_between(train_sizes, test_scores_mean - test_scores_std,
                 test_scores_mean + test_scores_std, alpha=0.1,
                 color="g")
plt.plot(train_sizes, train_scores_mean, 'o-', color="r",
         label="Training score")
plt.plot(train_sizes, test_scores_mean, 'o-', color="g",
         label="Cross-validation score")
plt.legend(loc="best")
# Determine if the model is underfitting, overfitting or balanced
if train_scores_mean[-1] < test_scores_mean[-1]:
    print('The model is underfitting.')
elif train_scores_mean[-1] > test_scores_mean[-1]:
    print('The model is overfitting.')
else:
    print('The model is balanced.')

plt.show()
```
